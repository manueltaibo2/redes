```python
# ej2.2
import smtplib
import time

mensaje = """\
Subject: prueba tls
From: from@example.com
To: to@example.com

con tls."""

with smtplib.SMTP("sandbox.smtp.mailtrap.io", 2525) as server:
    server.starttls()
    server.login("51fe75e9fc9230", "21d121745b4ba8")
    server.sendmail("from@example.com", "to@example.com", mensaje)
    print("enviado con tls")
```

```python
#ej3.1
import poplib

server = poplib.POP3("pop3.mailtrap.io", 1100)
server.user("51fe75e9fc9230")
server.pass_("21d121745b4ba8")

num_correos = len(server.list()[1])
print(f"numero de correos: {num_correos}")

if num_correos > 0:
    mensaje = server.retr(num_correos)[1]
    for linea in mensaje:
        print(linea.decode("utf-8"))

server.quit()
```

```python
# ej3.2
import poplib
import ssl

contexto = ssl.create_default_context()

server = poplib.POP3("pop3.mailtrap.io", 9950)
server.stls(context=contexto)
server.user("51fe75e9fc9230")
server.pass_("21d121745b4ba8")

num_correos = len(server.list()[1])
print(f"numero de correos: {num_correos}")

if num_correos > 0:
    mensaje = server.retr(num_correos)[1]
    for linea in mensaje:
        print(linea.decode("utf-8"))

server.quit()
```
